```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=sns.load_dataset('titanic')
df=pd.DataFrame(data)
df.head(3)
```

```python
mo=df['fare'].mode()[0]
```

```python
# df['fare'].value_counts()
```

```python
df['age'].fillna(df['age'].mean(),inplace=True)
```

```python
md=np.median(df["fare"])
```

```python
df["fare"].median()
```

```python
df['fare'].mean()
```

```python
mm=np.mean(df['fare'])
```

```python
sns.histplot(x='fare',data=df,bins=[i for i in range(0,81,10)])
plt.plot([mm for i in range(0,300)],[i for i in range(0,300)],c='green',label='mean')
plt.plot([md for i in range(0,300)],[i for i in range(0,300)],c='blue',label='median')
plt.plot([mo for i in range(0,300)],[i for i in range(0,300)],c='red',label='mode')
plt.legend()
# plt.plot()
plt.show()
```

```python


import matplotlib.pyplot as plt
import numpy as np
from numpy.random import normal
from scipy.stats import norm
import seaborn as sns

sample = normal(loc=50, scale=5, size=1000)

# Calculate sample mean and standard deviation
sample_mean = sample.mean()
sample_std = sample.std()

# Fit normal distribution using sample parameters
dist = norm(sample_mean, sample_std)

# Generate x-values across the complete sample range
values = np.linspace(sample.min(), sample.max(), 100)

probability_density = dist.pdf(values)

# plt.hist(sample, bins=10, density=True)
# plt.plot(values, probability_density, color="red", linewidth=2)
# plt.xlabel("Values")
# plt.ylabel("Density")
# plt.title("Normal Distribution: Histogram + PDF")
# plt.show()
sns.distplot(sample, hist=True, kde=True)
plt.show()
```
