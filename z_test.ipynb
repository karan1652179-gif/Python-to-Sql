```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st

# Given values
s_x = 90          # Sample mean
p_u = 82          # Population mean
p_std = 20        # Population standard deviation
n = 81            # Sample size
ap = 0.05         # Significance level

# Calculate Z-statistic
z_cal = (s_x - p_u) / (p_std / np.sqrt(n))

# Calculate critical Z-value
z_table = st.norm.ppf(1 - ap)

# Print results
print("Calculated Z-value:", z_cal)
print("Critical Z-value:", z_table)

# Generate normal distribution
x = np.linspace(-4, 4, 1000)
y = st.norm.pdf(x)

# Plot graph
plt.figure(figsize=(12, 6))
plt.plot(x, y, label="Standard Normal Distribution")

# Shade rejection region
x_reject = np.linspace(z_table, 4, 300)
y_reject = st.norm.pdf(x_reject)
plt.fill_between(
    x_reject, y_reject, color="red", alpha=0.4,
    label="Rejection Region (α = 0.05)"
)

# Critical Z-value
plt.axvline(
    z_table, color="red", linestyle="--",
    label=f"Critical Z = {z_table:.3f}"
)

# Calculated Z-value
plt.axvline(
    z_cal, color="green", linestyle="-",
    label=f"Calculated Z = {z_cal:.2f}"
)

# Labels and title
plt.title("Right-Tailed Z-Test (α = 0.05)")
plt.xlabel("Z-value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Hypothesis decision
if z_cal > z_table:
    print("Reject H0: Ha is supported.")
else:
    print("Fail to reject H0: Insufficient evidence for Ha.")
```

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

# Step 1: Data
old_design_data = np.array([
    45.2, 42.8, 38.9, 43.5, 41.0, 44.6, 40.5, 42.7,
    39.8, 41.4, 44.3, 39.7, 42.1, 40.6, 43.0, 42.2,
    41.5, 39.6, 44.0, 43.1, 38.7, 43.9, 42.0, 41.9,
    42.8, 43.7, 41.3, 40.9, 42.5, 41.6
])

new_design_data = np.array([
    48.5, 49.1, 50.2, 47.8, 48.7, 49.9, 48.0, 50.5,
    49.8, 49.6, 48.2, 48.9, 49.7, 50.3, 49.4, 50.1,
    48.6, 48.3, 49.0, 50.0, 48.4, 49.3, 49.5, 48.8,
    50.6, 50.4, 48.1, 49.2, 50.7, 50.8
])

# Step 2: Given values
pop_std = 2.5
n_sp = len(new_design_data)
alpha = 0.05

# Step 3: Calculate means
mean_new = np.mean(new_design_data)
mean_old = np.mean(old_design_data)

print("Old Design Mean:", mean_old)
print("New Design Mean:", mean_new)

# Step 4: Calculate Z-statistic
z_cal = (mean_new - mean_old) / (pop_std / np.sqrt(n_sp))

# Step 5: Critical Z-value
z_table = st.norm.ppf(1 - alpha)

# Step 6: P-value
p_value = 1 - st.norm.cdf(z_cal)

print("Calculated Z-value:", z_cal)
print("Critical Z-value:", z_table)
print("P-value:", p_value)

# Step 7: Plot normal distribution
x = np.linspace(-4, 4, 1000)
y = st.norm.pdf(x)

plt.figure(figsize=(12, 6))

plt.plot(x, y, label="Standard Normal Distribution")

# Shade rejection region
x_reject = np.linspace(z_table, 4, 300)

plt.fill_between(
    x_reject,
    st.norm.pdf(x_reject),
    alpha=0.4,
    label="Rejection Region (alpha = 0.05)"
)

# Critical Z-value
plt.axvline(
    z_table,
    linestyle="--",
    label=f"Critical Z = {z_table:.3f}"
)

# Calculated Z-value
plt.axvline(
    z_cal,
    linestyle="-",
    label=f"Calculated Z = {z_cal:.2f}"
)

plt.title("Z-Test: Old Design vs New Design")
plt.xlabel("Z-value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Step 8: Hypothesis decision
if z_cal > z_table:
    print("Reject H0: Ha is supported.")
else:
    print("Fail to reject H0.")
```
