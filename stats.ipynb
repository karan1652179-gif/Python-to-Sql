```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips= sns.load_dataset('tips')
tips.head(3)

```

```python
tips.info()
```

```python
# calculate Correlation
data1=tips.select_dtypes(['float64','int64']).corr()
data1
```

```python
data2=tips.select_dtypes(['float64','int64']).cov()
data2
```

```python
sns.heatmap(data1,annot=True)
plt.show()
```

```python
sns.heatmap(data2,annot=True)
plt.show()
```

```python
# CENTRAL LIMIT THEOREM
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
pop_data=[np.random.randint(10,100) for i in range(10000)]
pop_table=pd.DataFrame({"pop_data":pop_data})
```

```python
plt.Figure(figsize=(3,2))
sns.kdeplot(x='pop_data',data=pop_table)
plt.show()
```

```python
sample_mean=[]
for i in range(50):
   sample_data=[]
   for data  in range(500):
      sample_data.append(np.random.choice(pop_data))
   sample_mean.append(np.mean(sample_data))
```

```python
sample_M=pd.DataFrame({'sample_mean':sample_mean})
sns.kdeplot(x='sample_mean',data=sample_M)
plt.show()
```
