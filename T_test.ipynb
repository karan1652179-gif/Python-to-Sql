# EXAMPLE OF POTATO CHIPS MANUFACTURE COMPANY:

# A manufacturer claims that the average weight of a bag of potato chips is 150
# grams. A sample of 25 bags is taken, and the average weight is found to be 148
# grams, with a standard deviation of 5 grams. Test the manufacturer's claim using a
# one-tailed t-test with a significance level of 0.05. 


```python
import scipy.stats as st
t_table= st.t.ppf(0.05,24)
t_table
```

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt


sample_mean = 148
population_mean = 150
sample_std = 5
n = 25
alpha = 0.05
 
df = n - 1

t_cal = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
t_table = st.t.ppf(alpha, df)
p_value = st.t.cdf(t_cal, df)


print("Calculated T-value:", t_cal)
print("Critical T-value:", t_table)
print("P-value:", p_value)


x = np.linspace(-4, 4, 1000)
y = st.t.pdf(x, df)

plt.figure(figsize=(12, 6))

plt.plot(x, y, label=f"T-Distribution (df = {df})")


x_reject = np.linspace(-4, t_table, 300)
plt.fill_between(
    x_reject,
    st.t.pdf(x_reject, df),
    alpha=0.4,
    label="Rejection Region (alpha = 0.05)"
)

# Critical t-value
plt.axvline(
    t_table,
    linestyle="--",
    label=f"Critical T = {t_table:.3f}"
)

# Calculated t-value
plt.axvline(
    t_cal,
    linestyle="-",
    label=f"Calculated T = {t_cal:.2f}"
)

plt.title("One-Sample T-Test: Potato Chips Weight")
plt.xlabel("T-value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Step 7: Hypothesis decision
if t_cal < t_table:
    print("Reject H0: Evidence supports Ha.")
else:
    print("Fail to reject H0: Insufficient evidence for Ha.")
```

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

# Example: Compare productivity between two teams
# Team A: Mean = 80, Standard deviation = 5
# Team B: Mean = 75, Standard deviation = 6
# Sample size = 20 for each team
# Significance level = 0.05

# Step 1: Given data
mean_a = 80
mean_b = 75

std_a = 5
std_b = 6

n_a = 20
n_b = 20

alpha = 0.05

# Step 2: Calculate degrees of freedom
df = n_a + n_b - 2

# Step 3: Calculate t-statistic
t_cal = (mean_a - mean_b) / np.sqrt(
    (std_a**2 / n_a) + (std_b**2 / n_b)
)

# Step 4: Calculate critical t-values (two-tailed test)
t_table = st.t.ppf(1 - alpha / 2, df)

# Step 5: Calculate p-value
p_value = 2 * st.t.sf(abs(t_cal), df)

# Step 6: Print results
print("Calculated T-value:", t_cal)
print("Critical T-value:", t_table)
print("Degrees of Freedom:", df)
print("P-value:", p_value)

# Step 7: Generate t-distribution
x = np.linspace(-4, 4, 1000)
y = st.t.pdf(x, df)

plt.figure(figsize=(12, 6))

plt.plot(x, y, label=f"T-Distribution (df={df})")

# Left rejection region
x_left = np.linspace(-4, -t_table, 300)
plt.fill_between(
    x_left,
    st.t.pdf(x_left, df),
    alpha=0.4,
    label="Rejection Regions (alpha=0.05)"
)

# Right rejection region
x_right = np.linspace(t_table, 4, 300)
plt.fill_between(
    x_right,
    st.t.pdf(x_right, df),
    alpha=0.4
)

# Critical values
plt.axvline(
    -t_table,
    linestyle="--",
    label=f"Critical T = ±{t_table:.3f}"
)

plt.axvline(
    t_table,
    linestyle="--"
)

# Calculated t-value
plt.axvline(
    t_cal,
    linestyle="-",
    label=f"Calculated T = {t_cal:.3f}"
)

plt.title("Independent Two-Sample T-Test")
plt.xlabel("T-value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Step 8: Hypothesis decision
if abs(t_cal) > t_table:
    print("Reject H0: Significant difference between teams.")
else:
    print("Fail to reject H0: No significant difference found.")
```

# Chi-Square test:
# Expample 1:
# A fair die is rolled 120 times and the following result are obtained:
# Face 1:22 times
# Face 2:17 times
# Face 3: 20 times
# face 4:26 times
# face 5:22 times
# face 6; 13 times
# Test at a 5% level of signifficance wether the die is fair.

```python

import numpy as np
ob=np.array([22,17,20,26,22,13])
ex=np.array([20,20,20,20,20,20])
np.sum(np.square(ob-ex)/ex)
```

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

# Chi-Square Test
# A fair die is rolled 120 times.

# Step 1: Observed frequencies
ob = np.array([22, 17, 20, 26, 22, 13])

# Step 2: Expected frequencies for a fair die
ex = np.array([20, 20, 20, 20, 20, 20])

# Step 3: Significance level
alpha = 0.05

# Step 4: Degrees of freedom
df = len(ob) - 1

# Step 5: Calculate Chi-Square statistic
chi_cal = np.sum((ob - ex)**2 / ex)

# Step 6: Calculate critical value
chi_table = st.chi2.ppf(1 - alpha, df)

# Step 7: Calculate p-value
p_value = st.chi2.sf(chi_cal, df)

# Step 8: Print results
print("Observed Frequencies:", ob)
print("Expected Frequencies:", ex)
print("Degrees of Freedom:", df)
print("Calculated Chi-Square:", chi_cal)
print("Critical Chi-Square:", chi_table)
print("P-value:", p_value)

# Step 9: Plot Chi-Square distribution
x = np.linspace(0, 20, 1000)
y = st.chi2.pdf(x, df)

plt.figure(figsize=(12, 6))

plt.plot(x, y, label=f"Chi-Square Distribution (df={df})")

# Shade rejection region
x_reject = np.linspace(chi_table, 20, 300)
plt.fill_between(
    x_reject,
    st.chi2.pdf(x_reject, df),
    alpha=0.4,
    label="Rejection Region (alpha = 0.05)"
)

# Critical value
plt.axvline(
    chi_table,
    linestyle="--",
    label=f"Critical Value = {chi_table:.3f}"
)

# Calculated value
plt.axvline(
    chi_cal,
    linestyle="-",
    label=f"Calculated Value = {chi_cal:.2f}"
)

plt.title("Chi-Square Goodness-of-Fit Test: Fair Die")
plt.xlabel("Chi-Square Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Step 10: Hypothesis decision
if chi_cal > chi_table:
    print("Reject H0: The die may not be fair.")
else:
    print("Fail to reject H0: No sufficient evidence that the die is unfair.")
```

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt

# Example 2: Chi-Square Test of Independence
# Study the relationship between gender and music preference.
# Total sample size = 235
# Significance level = 0.05

# Step 1: Observed frequencies
row1 = np.array([40, 45, 25, 10])
row2 = np.array([35, 30, 20, 30])

ob = np.array([row1, row2])

print("Observed Frequencies:")
print(ob)

# Step 2: Calculate row totals and column totals
sum_row = np.sum(ob, axis=1)
sum_col = np.sum(ob, axis=0)
total = np.sum(ob)

print("\nRow Totals:", sum_row)
print("Column Totals:", sum_col)
print("Grand Total:", total)

# Step 3: Calculate expected frequencies
exp = np.outer(sum_row, sum_col) / total

print("\nExpected Frequencies:")
print(exp)

# Step 4: Calculate Chi-Square statistic
chi_cal = np.sum((ob - exp)**2 / exp)

# Step 5: Calculate degrees of freedom
df = (ob.shape[0] - 1) * (ob.shape[1] - 1)

# Step 6: Calculate critical value
alpha = 0.05
chi_table = st.chi2.ppf(1 - alpha, df)

# Step 7: Calculate p-value
p_value = st.chi2.sf(chi_cal, df)

# Step 8: Display results
print("\nCalculated Chi-Square:", chi_cal)
print("Critical Chi-Square:", chi_table)
print("Degrees of Freedom:", df)
print("P-value:", p_value)

# Step 9: Plot Chi-Square distribution
x = np.linspace(0, 20, 1000)
y = st.chi2.pdf(x, df)

plt.figure(figsize=(12, 6))

plt.plot(x, y, label=f"Chi-Square Distribution (df={df})")

# Shade rejection region
x_reject = np.linspace(chi_table, 20, 300)
plt.fill_between(
    x_reject,
    st.chi2.pdf(x_reject, df),
    alpha=0.4,
    label="Rejection Region (alpha = 0.05)"
)

# Critical value
plt.axvline(
    chi_table,
    linestyle="--",
    label=f"Critical Value = {chi_table:.3f}"
)

# Calculated value
plt.axvline(
    chi_cal,
    linestyle="-",
    label=f"Calculated Value = {chi_cal:.3f}"
)

plt.title("Chi-Square Test of Independence")
plt.xlabel("Chi-Square Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Step 10: Hypothesis decision
if chi_cal > chi_table:
    print("\nReject H0: Gender and music preference are associated.")
else:
    print("\nFail to reject H0: No significant association was found.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
mu = 260
sigma = 31
alpha = 0.05

# Critical z-values for a two-tailed test
z_critical = norm.ppf(1 - alpha / 2)

lower = mu - z_critical * sigma
upper = mu + z_critical * sigma

# Generate normal distribution
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = norm.pdf(x, mu, sigma)

# Plot
plt.figure(figsize=(12, 6))

plt.plot(x, y, color="black", linewidth=2,
         label="Normal Distribution")

# Left rejection region
x_left = x[x <= lower]
plt.fill_between(
    x_left, norm.pdf(x_left, mu, sigma),
    color="red", alpha=0.6, label="Rejection Region"
)

# Right rejection region
x_right = x[x >= upper]
plt.fill_between(
    x_right, norm.pdf(x_right, mu, sigma),
    color="red", alpha=0.6
)

# Critical values
plt.axvline(lower, color="red", linestyle="--",
            label=f"Lower Critical Value = {lower:.2f}")
plt.axvline(upper, color="red", linestyle="--",
            label=f"Upper Critical Value = {upper:.2f}")

# Mean
plt.axvline(mu, color="blue", linestyle="-.",
            label=f"Mean = {mu}")

# Alpha labels
plt.text(lower - 15, 0.0015, "α/2 = 0.025",
         ha="center", color="red")
plt.text(upper + 15, 0.0015, "α/2 = 0.025",
         ha="center", color="red")

# Formatting
plt.title("Steps of Hypothesis Testing: Two-Tailed Test",
          fontsize=16, fontweight="bold")
plt.xlabel("Energy Cost")
plt.ylabel("Probability Density")

plt.legend(loc="upper left", fontsize=9)
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```
